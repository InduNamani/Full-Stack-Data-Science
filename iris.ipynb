```python
import pandas as pd

df = pd.read_csv(r'C:\Users\indun\Downloads\archive (1)\iris.csv')

# Check current names
print("Original columns:", df.columns.tolist())

# Rename columns (adjust based on what you see)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# OR if there's an Id column:
# df.columns = ['id', 'sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
# df = df.drop('id', axis=1)  # Remove Id column

print("New columns:", df.columns.tolist())
print(df.head())
```

# Matplotlib

```python
plt.figure(figsize=(8,4))
s_c=df['species'].value_counts()
plt.bar(s_c.index,s_c.values,color=['red','green','blue'])
plt.title('Number of samples per species',fontsize=14)
plt.xlabel('Species')
plt.ylabel('count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,6))
colors={'setosa':'red','versicolor':'green','virginica':'blue'}
for i in df['species'].unique():
    s_d=df[df['species']==i]
    plt.scatter(s_d['sepal_length'],
                s_d['sepal_width'],
                label=i,
                color=colors[i],
                alpha=0.6,
                s=100)
plt.title('sepal length vs sepal width ')
plt.xlabel('sepal length')
plt.ylabel('sepal width')
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```

```python
# Line plot
plt.figure(figsize=(10, 6))
avg_values = df.groupby('species')[['sepal_length', 'sepal_width', 
                                      'petal_length', 'petal_width']].mean()
avg_values.T.plot(marker='o', linewidth=2, markersize=8)
plt.title('Average Measurements by Species', fontsize=14)
plt.xlabel('Features', fontsize=12)
plt.ylabel('Average Value (cm)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Species')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

# seaborn visualization

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='species', palette='Set2')
plt.title('Species Distribution', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()
```

```python
# Box Plot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='species', y='petal_length', palette='viridis')
plt.title('Petal Length Distribution by Species', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.show()
```

```python
# voilin plot
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='species', y='sepal_width', palette='muted')
plt.title('Sepal Width Distribution by Species', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.show()
```
